This code is to demonstrate how to generate the Mixed Higher Order Complex of the product of an arbitrary number of chains. This is meant to be run as a Jupyter Notebook with a Sage 10.6 Kernel. For all examples r=2.

```python
#### Importing all the relevant things
import detideal
import higher_order_complex
```

```python
##### First make the higher order complex for [3]x[2]x[3] and print its hvector
MixedHigherOrderComplex323 = higher_order_complex.higher_order_complex([3,2,3],2)
print(MixedHigherOrderComplex323.hvector())
#### Now make the determinental ideal of r x r minors for unfoldings of the 3 tensor and print the hilbert series numerator
R323,I323 =  detideal.make_ideal([3,2,3], 2)
num323 = I323.hilbert_series().numerator()
print(num323)
```

```python
##### First make the higher order complex for [3]x[2]x[2]x[3] and print its hvector
MixedHigherOrderComplex3223 = higher_order_complex.higher_order_complex([3,2,2,3],2)
print(MixedHigherOrderComplex3223.hvector())
#### Now make the determinental ideal of r x r minors for unfoldings of the 4 tensor and print the hilbert series numerator
R3223,I3223 =  detideal.make_ideal([3,2,2,3], 2)
num3223 = I3223.hilbert_series().numerator()
print(num3223)
```
